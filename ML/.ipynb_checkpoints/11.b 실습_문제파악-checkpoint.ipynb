### 기술 지원 데이터는 종종 고객 경험 개선 기회에 대한 풍부한 정보 소스가 될 수 있습니다. 제품에 대한 고객의 문제가 적을수록 좋습니다. 고객이 최소한의 노력으로 기술적 과제를 신속하게 극복 할 수 있다면 더욱 좋습니다. 기술 지원 데이터를 분석하고 문제 유형, 문제 해결 시간 및 가장 적합한 지원 채널에 대한 기본적인 분석을 수행하겠습니다.

## technical_support_data.csv 파일을 통해, 비슷한 유형의 문제들을 도출하시오.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/technical_support_data.csv')
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
X = df.loc[ : , 'no_of_cases' : 'Post_warranty_percent']
```

```python
X.head(2)
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
for k in range(1,11) :
    kmeans = KMeans(n_clusters = k, random_state = 5)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
df_wcss = pd.Series(wcss, index= range(1,10+1))
```

```python
df_wcss.plot(kind = 'bar')
plt.show()
```

```python
kmeans = KMeans(n_clusters = 5, random_state = 10)
```

```python
y_pred = kmeans.fit_predict(X)
```

```python
df['Group'] = y_pred
```

```python
df.head()
```

```python
df.loc[ df['Group']== 2, ]
```

```python
df = pd.read_csv('../data/food_delivery_customer_dataset_original.csv')
```

```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, StandardScaler ,MinMaxScaler
```

```python
from sklearn.compose import ColumnTransformer
```

```python
ct = ColumnTransformer([('onehot', OneHotEncoder(), [4,5]),
                         ('scaler', MinMaxScaler(), [0,1,2,3,6])])
```

```python
X = ct.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
for k in range(1,11) :
    kmeans = KMeans(n_clusters = k, random_state = 5)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
df_wcss = pd.Series(wcss, index= range(1,10+1))
```

```python
df_wcss.plot( )
plt.show()
```

```python
df_wcss.plot(kind = 'bar')
plt.show()
```

```python
kmeans = KMeans(n_clusters = 5, random_state = 10)
```

```python
y_pred = kmeans.fit_predict(X)
```

```python
df['Group'] = y_pred
```

### => 'region'으로 나눠져버림 => 클러스터링 하는 의미가 없다

#### 1. region 컬럼을 제외하고 클러스터링 하기

#### 2. silhouette 실루엣 스코어를 계산한다

```python
from sklearn.metrics import silhouette_score
```

```python
score_list = []
for k in range(2, 10+1) :   # => 실루엣 스코어 구할 때 , 클러스터는 2개 이상으로 넣어줘야함 
    kmeans = KMeans(n_clusters = k, random_state = 5)
    y_pred = kmeans.fit_predict(X)
    score = silhouette_score( X, y_pred )
    score_list.append(score)
```

### silhouette score는  
=> -1~1 사이의 숫자이고, 1에 가까울수록 군집이 잘 뭉쳐있는 것이다

```python
score_list
```

```python
X = df.drop(['customer_id','region'], axis = 1)
```

```python
ct = ColumnTransformer([('onehot', OneHotEncoder(), [4]),('scaler', MinMaxScaler(), [0,1,2,3,5])])
```

```python
X.head(2)
```

```python
X = ct.fit_transform(X)
```

```python
wcss = []
score_list = []
for k in range(2,11) :
    kmeans = KMeans(n_clusters = k, random_state = 5)
    y_pred = kmeans.fit_predict(X)
    score = silhouette_score(X,y_pred)
    score_list.append(score)
    wcss.append(kmeans.inertia_)  
```

```python
df_wcss = pd.Series(wcss, index= range(2,10+1))
```

```python
score_list
```

```python
df_wcss.plot()
plt.show()
```

```python
df_wcss.plot(kind = 'bar')
plt.show()
```

```python
kmeans = KMeans(n_clusters = 5, random_state = 10)
```

```python
y_pred = kmeans.fit_predict(X)
```

```python
df['Group'] = y_pred
```
